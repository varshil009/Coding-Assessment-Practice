#### 1. Valid Paranthesis

```python
s1 = "()"
# Expected Output: True
s2 = "()[]{}"
# Expected Output: True
s3 = "(]"
# Expected Output: False
s4 = "]"
# Expected Output: False
```

```python
def valid_p(stack):
    brackets = {"}" : "{", ")" : "(", "]" : "["}
    stackx = []
    for s in stack:
        if s in "([{":
            stackx.append(s)
        else:
            if not stackx:
                return False
                
            if stackx[-1] == brackets[s]:
                stackx.pop()
            else:
                return False
    
    return not stackx
```

```python
print(valid_p(s1))
print(valid_p(s2))
print(valid_p(s3))
print(valid_p(s4))
```

#### 2. Daily Temp

```python
# Test Case 1
temperatures1 = [73, 74, 75, 71, 69, 72, 76, 73]
# Expected Output: [1, 1, 4, 2, 1, 1, 0, 0]

# Test Case 2
temperatures2 = [30, 40, 50, 60]
# Expected Output: [1, 1, 1, 0]

# Test Case 3
temperatures3 = [30, 60, 90]
# Expected Output: [1, 1, 0]
```

```python
def temp(temps):
    l = len(temps)
    answer = [0] * l
    stack = []
    for i, t in enumerate(temps):
        # While the current temperature is warmer than the temperature 
        # recorded at the top of the stack:
        while stack and t > temps[stack[-1]]:
            prev_index = stack.pop()
            answer[prev_index] = i - prev_index
        stack.append(i)

    return answer
```

```python
print(temp(temperatures1))
print(temp(temperatures2))
print(temp(temperatures3))
```

```python
def binary_search(nums, target):
    left, right = 0, len(nums) - 1
    while left <= right:
        mid = left + (right - left) // 2
        if nums[mid] == target:
            return mid

        elif nums[mid] < target:
            left = mid + 1
        
        elif nums[mid] > target:
            right = mid - 1
    return -1
```

```python
def searchMatrix(matrix: list[list[int]], target: int) -> bool:
    if not matrix or not matrix[0]:
        return False
        
    rows, cols = len(matrix), len(matrix[0])
    # flatlined indexes
    left, right = 0, (rows * cols) - 1
    
    while left <= right:
        mid = left + (right - left) // 2
        
        # Virtual 2D Index Mapping
        r, c = mid // cols, mid % cols
        val = matrix[r][c]
        
        if val == target:
            return True
        elif val < target:
            left = mid + 1
        else:
            right = mid - 1
            
    return False
```

```python
# find minimum value in a rotated matrix
def findMin(nums: list[int]) -> int:
    left, right = 0, len(nums) - 1
    
    while left < right:
        mid = left + (right - left) // 2
        
        # If mid value is greater than rightmost value,
        # the minimum MUST be in the right half!
        if nums[mid] > nums[right]:
            left = mid + 1
        # Otherwise, minimum is at mid or in the left half
        else:
            right = mid
            
    # When left == right, we've pinpointed the minimum!
    return nums[left]
```

```python
# Test Case 1
prices1 = [7, 1, 5, 3, 6, 4]
# Expected Output: 5
# Explanation: Buy on day 2 (price = 1) and sell on day 5 (price = 6), profit = 6 - 1 = 5.

# Test Case 2
prices2 = [7, 6, 4, 3, 1]
# Expected Output: 0
# Explanation: In this case, no transactions are done and max profit = 0.
```

```python
def buynsell(arr):
    min_val = float('inf')
    max_profit = float('-inf')
    for x in arr:
        if x < min_val:
            min_val = x
        
        potential_profit = x - min_val

        if potential_profit > max_profit:
            max_profit = potential_profit
    
    return max_profit
```

```python
print(buynsell(prices1))
print(buynsell(prices2))
```

```python
def buynsellx(arr):
    left, right = 0, len(arr)-1
    max_val, min_val = float('-inf'), float('inf')
    for i, x in enumerate(arr):
        if x > max_val and right > left and left >= i:
            max_val = x
            right = i

        if x < min_val and left < right and right >= i:
            min_val = x
            left = i

    return arr[right] - arr[left]
```

```python
print(buynsell(prices1))
print(buynsell(prices2))
```

```python
def buynsell_sliding_window(prices: list[int]) -> int:
    left = 0   # Buy day
    right = 1  # Sell day
    max_profit = 0

    while right < len(prices):
        # Is the transaction profitable?
        if prices[left] < prices[right]:
            profit = prices[right] - prices[left]
            max_profit = max(max_profit, profit)
        else:
            # We found a new lowest price! Move 'buy' pointer to 'sell'
            left = right
            
        right += 1  # Always advance 'sell' day forward

    return max_profit
```

```python
from collections import defaultdict
def substring_nonrepeating(strx):
    hashmap = defaultdict(int)
    subs = {}
    biggest = 0
    s = ""
    for x in strx:
        hashmap[x] +=1
        if hashmap[x] > 1:
            slen = len(s)
            subs[slen] = s
            if slen > biggest:
                biggest = slen
            s = f""
            hashmap = defaultdict(int)
            hashmap[x] +=1

        s += x

    return subs[biggest]
```

```python
print(substring_nonrepeating(s1))
print(substring_nonrepeating(s2))
print(substring_nonrepeating(s3))
```

```python
def lengthOfLongestSubstring(s: str) -> int:
    char_set = set()
    left = 0
    max_len = 0
    
    for right in range(len(s)):
        # While the incoming character is a duplicate, shrink the window from the left
        while s[right] in char_set:
            char_set.remove(s[left])
            left += 1
            
        # Add the new character and record the max window size
        char_set.add(s[right])
        max_len = max(max_len, right - left + 1)
        
    return max_len
```

```python
print(lengthOfLongestSubstring(s1))
print(lengthOfLongestSubstring(s2))
print(lengthOfLongestSubstring(s3))
```
