```python
def two_sum(nums: list[int], target: int) -> list[int]:
    """
    basic idea is, when we are on loop, save the iterated num and its index as hasmhamp;
    and we calculate difference between current num and target; if difference number is in hashmap, 
    which means its a counter part of current number which adds up to the target and we take that current 
    number index and access counter part index through a hashmap
    """
    seen = {}  # Map number -> index
    ls = []
    for current_index, num in enumerate(nums):
        diff = target - num
        # O(1) fast lookup
        if diff in seen:
            ls.append([seen[diff], current_index])

        # Save number and its index for later numbers to match with
        seen[num] = current_index
    return ls

two_sum(nums1, target1)
```

```python
print(two_sum([3, 3], 6))         # Output: [0, 1]  (Duplicates forming target)
print(two_sum([-3, 4, 3, 90], 0))  # Output: [0, 2]  (Negatives & Zero target)
print(two_sum([0, 4, 3, 0], 0))   # Output: [0, 3]  (Zeroes)
```

```python
def maxProfit(prices):
    """
    simply get current min value, subtract it with current value, update profit value
    """
    min_price = float('inf')
    max_profit = 0

    for x in prices:
        if x < min_price:
            min_price = x
        
        potential_profit = x - min_price

        if potential_profit > max_profit:
            max_profit = potential_profit
    
    return max_profit
```

```python
def removeDuplicates(nums: list[int]) -> int:
    if not nums:
        return 0
        
    # 'write' pointer starts at 1 because index 0 is always unique
    write = 1
    
    for read in range(1, len(nums)):
        # If the current number is different from the last unique number we placed:
        if nums[read] != nums[write - 1]:
            nums[write] = nums[read]
            write += 1
            
    return write  # Returns 'k' (the number of unique elements)
```

```python
def mergeIntervals(interval):

    """
    the solution is right for single interval merge but doesnt takes into account independent ones
    meaning you do have to return the intervals that are different; here simple logic is to check boundaries
    if x for currengt interval is between x and y of nominated ine=terval, then check its y
    same for y of current interbal, if it is between x and y of nominated interbal then check its x and update
    """
    def set_y(y, cy):
        if y >= cy:
            return y
        else:
            return cy

    def set_x(x, cx):
        if x <= cx:
            return x
        else:
            return cx

    cx, cy = interval[0]
    for x, y in interval:
        #print(x, y)
        try:
            if x >= cx and x <= cy:
                cy = set_y(y, cy)
            if y <= cy and y >= cx: 
                cx = set_x(x, cx)
        except Exception as e:
            print("error :", x, y)

    return cx, cy

```

```python
def mergeIntervals(interval):
    if not interval:
        return []

    interval.sort(key = lambda x : x[0])

    ints = [interval[0]]
    for x, y in interval[1:]:
        intx, inty = ints[-1][0], ints[-1][1]
        if x >= intx  and x <= inty:
            ints[-1][1] = max(y, inty)
        else:
            ints.append([x, y])
    
    return ints
    
```

```python
# alright the optimal solution is different; space : O(1) and time : o(n)
def palindrome2(s):
    """
    whithout using extra string, iterates thorugh left to right by increasing index fro left and from right
    nested while loops skips non alphanumeric strings and in end comparision dictates output
    """
    left = 0
    right = len(s) - 1
    while left < right:
        # while loop skips non alphanumeric numbers for both left and right and then compare
        while left < right and not s[left].isalnum():
            left += 1
        while left < right and not s[right].isalnum():
            right -= 1
        if s[left].lower() != s[right].lower():
            return False

        left += 1
        right -= 1

    return True
```

```python
def prefix(l):
    ls = len(l[0])
    smallest_str = l[0]
    for x in l[1:]:
        if len(x) < ls:
            smallest_str = x

    match = ""
    for i, c in enumerate(smallest_str):
        if all([s[i]==c for s in l]): 
            match += c
            continue
        else:
            return match

    return match
```

```python
# optimal and fastest;
def prefix2(list):
    if not list:
        return ""

    prefix = []
    for c in zip(*list):
        if len(set(c)) == 1:
            prefix.append(c[0])
        else: 
            break

    return "".join(prefix)
```

```python
import collections
def topK(nums, k):
    hmap = sorted(collections.Counter(nums).items(), key = lambda x : x[1], reverse = True)
    return [x[0] for x in hmap[:k]]
```

```python
import collections

def topKFrequent(nums: list[int], k: int) -> list[int]:
    """
    bucket index is a frequency of value and inside that indexed list actual value;
    causes most of buckets to be empty cuz of repetition of ints
    reverse iteration, on bucket length index, which is a frequency of nums element,
    nested iteration over that bucket (empty most of time or single element present),
    the reverse iteration because; since higher frequency element comes first on reverse index;
    return on K achieved
    """
    count = collections.Counter(nums)
    # Buckets where index = frequency (max frequency is len(nums))
    buckets = [[] for _ in range(len(nums) + 1)]
    
    for num, freq in count.items():
        buckets[freq].append(num)
    result = []
    # Traverse from highest frequency down to 1
    for freq in range(len(buckets) - 1, 0, -1):
        for num in buckets[freq]:
            result.append(num)
            if len(result) == k:
                return result
    return result
```

```python
from collections import defaultdict
def subarrays2(nums, k):
    """
    Prefix Sum + HashMap (O(n) time, O(n) space)

    Very similar idea to Two Sum:
    instead of storing previously seen numbers, we store previously seen
    prefix sums (cumulative sums) in a hashmap.

    Key   -> Prefix Sum
    Value -> Number of times this prefix sum has occurred

    At every element:
        current_sum += nums[i]

    To find whether there exists a subarray ending at the current index whose
    sum equals k, we look for:

        diff = current_sum - k

    If 'diff' exists in the hashmap, it means we have previously encountered
    a prefix sum equal to 'current_sum - k'. Removing everything before that
    previous prefix leaves a contiguous subarray whose sum is exactly k.

    Mathematically:

        Current Prefix Sum - Previous Prefix Sum = Subarray Sum

    or

        current_sum - previous_prefix = k

    Example:
    nums = [1,2,1], k = 3

    Prefix sums:
    0, 1, 3, 4

    At current_sum = 4,
    diff = 4 - 3 = 1

    Since prefix sum 1 already exists (after processing the first element),
    the subarray between those two prefixes is:

        [2,1]

    whose sum is 3.

    The hashmap stores counts instead of booleans because the same prefix sum
    can occur multiple times, meaning multiple valid subarrays may end at the
    current index.

    We initialize:

        prefix_counts[0] = 1

    to represent the empty prefix before the array starts, allowing subarrays
    that begin from index 0 to be counted.
    """
    prefix_counts = defaultdict(int)
    prefix_counts[0] = 1
    current_sum = 0
    count = 0
    
    for x in nums:
        current_sum +=x
        diff = current_sum - k
        if diff in prefix_counts:
            count += prefix_counts[diff]
        
        prefix_counts[current_sum] +=1
    return count
            
```

```python
print(subarrays2(nums1, k1))
print(subarrays2(nums2, k2))
print(subarrays2(nums3, k3))
```
