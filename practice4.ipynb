## Binary Tree

A Binary Tree is a hierarchical data structure where each node has at most two children, typically referred to as the left child and the right child.Unlike a sequential Linked List, a Binary Tree branches out, making it exceptionally fast for searching and organizing data.

```python
class TreeNode:
    def __init__(self, value):
        self.val = value
        self.left = None   # Pointer to left child
        self.right = None  # Pointer to right child
```

      1  <-- Root Node

  4      <-- Leaf Node (No children)


Root: The topmost node of the tree (e.g., 1). <br>
Parent / Child: 1 is the parent of 2 and 3.  <br>
Leaf: A node with no children (e.g., 4, 3).  <br>
Subtree: Any node treated as the root of its own smaller tree (e.g., node 2 and its child 4 form a left subtree)

```python
# inorder traverseral, used extensively in binary trees because it visits numbers in ascending order
# Left -> Root -> Right
def inorder(root):
    if root:
        inorder(root.left)   # Go Left
        print(root.val, end=" ")  # Visit Root
        inorder(root.right)  # Go Right


# preorder traversal, Perfect for cloning or copying a tree architecture.
# Root -> Left -> Right
def preorder(root):
    if root:
        print(root.val, end=" ")  # Visit Root
        preorder(root.left)  # Go Left
        preorder(root.right) # Go Right

# post order traversal, Great for deleting a tree or calculating bottom-up metrics (like the height of a tree), 
# because you process the children before the parent
# Left -> Right -> Root
def postorder(root):
    if root:
        postorder(root.left)   # Go Left
        postorder(root.right)  # Go Right
        print(root.val, end=" ")  # Visit Root

```

### making a BTREE

#### The Insertion Algorithm :
If the tree is empty, create a new node. <br>
If the new value is smaller than the current node, move to the left child. <br>
If the new value is larger than the current node, move to the right child. <br>
Repeat recursively until you find an empty slot <br>

```python
class TreeNode:
    def __init__(self, value):
        self.val = value
        self.left = None
        self.right = None

def insert(root, value):
    # Base Case: If the tree/subtree is empty, return a new node
    if not root:
        return TreeNode(value)
    
    # Recursive Directional Steps
    if value < root.val:
        root.left = insert(root.left, value)
    elif value > root.val:
        root.right = insert(root.right, value)
        
    return root

def build_bst(arr):
    if not arr:
        return None
    
    # Step 1: First element becomes the root
    root = TreeNode(arr[0])
    
    # Step 2: Insert remaining elements
    for value in arr[1:]:
        insert(root, value)
        
    return root
```

```python
def search_bst(root, target):
    # Base Cases: root is null or key is present at root
    if not root or root.val == target:
        return root
    
    # Target is smaller than root's key -> search left
    if target < root.val:
        return search_bst(root.left, target)
    
    # Target is greater than root's key -> search right
    return search_bst(root.right, target)
```

```python
def max_depth(root):
    if not root:
        return 0  # An empty tree has a height of 0
    
    left_height = max_depth(root.left)
    right_height = max_depth(root.right)
    
    return 1 + max(left_height, right_height)
```

```python
def find_min(root):
    if not root:
        return None
    
    current = root
    while current.left:
        current = current.left  # Keep sliding left
        
    return current.val
find_min(k)
```

```python
def get_max_depth(root):
    if not root:
        return 0

    left_depth = get_max_depth(root.left)
    right_depth = get_max_depth(root.right)

    return 1 + max(left_depth, right_depth)
```

```python
def invert_tree(root):
    if not root:
        return
    root.left, root.right = root.right, root.left
    invert_tree(root.left)
    invert_tree(root.right)

    return root
```

```python
# Tree 1 (Valid BST):
#       2
#      / \
#     1   3
root1 = TreeNode(2, TreeNode(1), TreeNode(3))
# Expected Output: True

# Tree 2 (INVALID BST):
#       5
#      / \
#     1   4
#        / \
#       3   6
root2 = TreeNode(5, TreeNode(1), TreeNode(4, TreeNode(3), TreeNode(6)))
# Expected Output: False 
# (Explanation: Root is 5, but right child 4 is smaller than 5!)

root3 = TreeNode(10, TreeNode(5), TreeNode(15, TreeNode(6), TreeNode(20)))
```

```python
def valid_BST(root):
    """
    only valid for local search, going inside this doesnt verify parent nodes
    """
    if not root:
        return None

    if root.left and root.left.val > root.val:
        return False
    else:
        dontgoleft = True

    if root.right and root.right.val < root.val:
        return False
    else:
        dontgoright = True
    
    right = True if dontgoright else valid_BST(root.right)
    left = True if dontgoleft else valid_BST(root.left)

    return all([right, left])
```

```python
print(valid_BST(root1))
print(valid_BST(root2))
print(valid_BST(root3))
```

```python
def isValidBST(root: TreeNode) -> bool:
    def validate(node, low=float('-inf'), high=float('inf')):
        if not node:
            return True
            
        # Node value must be strictly inside (low, high)
        if not (low < node.val < high):
            return False
            
        # Going Left: upper bound becomes node.val
        # Going Right: lower bound becomes node.val
        return (validate(node.left, low, node.val) and 
                validate(node.right, node.val, high))

    return validate(root)
```

```python
# Tree Structure:
#        6
#      /   \
#     2     8
#    / \   / \
#   0   4 7   9
#      / \
#     3   5

root = TreeNode(6)
root.left = TreeNode(2, TreeNode(0), TreeNode(4, TreeNode(3), TreeNode(5)))
root.right = TreeNode(8, TreeNode(7), TreeNode(9))

p1, q1 = root.left, root.right  # Nodes 2 and 8
# Expected Output: Node 6

p2, q2 = root.left, root.left.right  # Nodes 2 and 4
# Expected Output: Node 2 (Node can be ancestor of itself!)
```

```python
def lowest_common_ancestor(root, p, q, ancestory_line = set()):
    if root == p or root == q:
        return ancestory_line
    elif not root:
        return None
    else:
        ancestory_line.add(root.val)
        match_p = lowest_common_ancestor(root.left, p, q, ancestory_line)
        match_q = lowest_common_ancestor(root.right, p, q, ancestory_line)

        if not match_p or not match_q:
            return None
        return list(match_p & match_q)[-1]
    
```

```python
print(lowest_common_ancestor(root, p1, q1))
print(lowest_common_ancestor(root, p2, q2))
```

```python
def lowestCommonAncestor(root: TreeNode, p: TreeNode, q: TreeNode) -> TreeNode:
    curr = root
    
    while curr:
        # Both values are smaller -> LCA is in left subtree
        if p.val < curr.val and q.val < curr.val:
            curr = curr.left
        # Both values are larger -> LCA is in right subtree
        elif p.val > curr.val and q.val > curr.val:
            curr = curr.right
        # Split point reached! Current node is the LCA
        else:
            return curr
```
