```python
class Node:
    def __init__(self, data):
        self.data = data  # Stores the actual value
        self.next = None  # Pointer to the next node

class LinkedList:
    def __init__(self):
        self.head = None  # Entry point of the linked list
    
    def traverse(self):
        current = self.head
        while current:
            print(current.data, end=" -> ")
            current = current.next
        print("None")

    def insert_at_head(self, new_data):
        new_node = Node(new_data)
        new_node.next = self.head  # Point new node to current head
        self.head = new_node       # Move head to point to the new node
    
    def insert_at_tail(self, new_data):
        new_node = Node(new_data)
        if not self.head:          # If list is empty
            self.head = new_node
            return
        
        current = self.head
        while current.next:        # Traverse until the last node
            current = current.next
        current.next = new_node    # Link the old last node to the new node

    def delete_node(self, key):
        current = self.head

        # Case 1: The head node itself holds the key to be deleted
        if current and current.data == key:
            self.head = current.next
            current = None
            return

        # Case 2: Search for the key to be deleted, keep track of the previous node
        prev = None
        while current and current.data != key:
            prev = current
            current = current.next
        
        # If key was not present in the list
        if not current:
            return

        # Unlink the node from the linked list
        prev.next = current.next
        current = None

    def search(self, target):
        current = self.head
        while current:
            if current.data == target:
                return True        # Found the target
            current = current.next
        return False               # Target not in list
```

```python
# Initialize list
ll = LinkedList()

# Insert data
ll.insert_at_head(10)
ll.insert_at_head(20)
ll.insert_at_tail(30)
# List is now: 20 -> 10 -> 30 -> None

print("Original List:")
ll.traverse() 

# Search data
print("Is 10 in list?", ll.search(10)) # True

# Delete data
ll.delete_node(10)
print("After deleting 10:")
ll.traverse() # Output: 20 -> 30 -> None
```

```python

def mergeTwoLists(l1, l2):
    # 1. Create a dummy sentinel node and a tail pointer pointing to it
    dummy = Node(0)
    tail = dummy

    # 2. Compare and append the smaller node
    while l1 and l2:
        if l1.value <= l2.value:
            tail.next = l1
            l1 = l1.next
        else:
            tail.next = l2
            l2 = l2.next
        
        # Advance the tail of our merged list
        tail = tail.next

    # 3. Attach whatever remains in l1 or l2
    tail.next = l1 if l1 else l2

    # Return the actual head (skipping dummy)
    return dummy.next
```

### linked list cycle

```python
Test Case 1:<br>
3 -> 2 -> 0 -> -4 
     ^          |   (Node -4 points back to Node 2)<br>
     |----------|
Expected Output: True

Test Case 2:
1 -> 2
^    |              (Node 2 points back to Node 1)<br>
|----|
Expected Output: True

Test Case 3:
1 -> None
Expected Output: False
```

```python
class ListNode:
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

def build_linked_list_with_cycle(values: list[int], pos: int) -> ListNode:
    """Helper: Builds a linked list and connects the tail to index `pos` (-1 for no cycle)."""
    if not values:
        return None
    
    nodes = [ListNode(val) for val in values]
    for i in range(len(nodes) - 1):
        nodes[i].next = nodes[i + 1]
        
    if pos != -1:
        nodes[-1].next = nodes[pos]  # Create cycle
        
    return nodes[0]

# --- TEST CASES ---

# Test Case 1: Cycle exists at index 1 (3 -> 2 -> 0 -> -4 -> 2 ...)
head1 = build_linked_list_with_cycle([3, 2, 0, -4], pos=1)
# Expected Output: True

# Test Case 2: Cycle exists at index 0 (1 -> 2 -> 1 ...)
head2 = build_linked_list_with_cycle([1, 2], pos=0)
# Expected Output: True

# Test Case 3: No cycle (1 -> None)
head3 = build_linked_list_with_cycle([1], pos=-1)
# Expected Output: False
```

```python
def circular_ll(head):
    history = set()
    current = head
    while current:
        if current not in history:
            history.add(current)
        else:
            return True
        current = current.next
    
    return False
```

```python
print(circular_ll(head1))
print(circular_ll(head2))
print(circular_ll(head3))
```

```python
def circular_llx(head):
    """
    If two runners are on a straight track, the faster runner just disappears ahead. 
    But if they run on a circular loop, the faster runner will eventually lap the 
    slower runner and collide from behind!
    """
    slow, fast = head, head
    while fast and fast.next:
        slow = slow.next
        fast = fast.next.next
        if slow == fast:
            return True
    
    return False
```

```python
print(circular_llx(head1))
print(circular_llx(head2))
print(circular_llx(head3))
```
